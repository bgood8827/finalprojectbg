```python
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
```

```python
michael_jordan_regular_stats = pd.read_csv("/Users/bradygoodmandesk/Desktop/jordannormal2.csv")
michael_jordan_total_stats = pd.read_csv("/Users/bradygoodmandesk/Desktop/jordantotals.csv")
michael_jordan_advanced_stats = pd.read_csv("/Users/bradygoodmandesk/Desktop/jordanadvanced.csv")
michael_jordan_regular_stats
michael_jordan_total_stats
michael_jordan_advanced_stats
```

```python
michael_jordan_selected_regular_stats = michael_jordan_regular_stats[['Season', 'G', 'GS', 'PTS', 'TRB', 'AST', 'STL', 'BLK', 'FG%']]
michael_jordan_selected_advanced_stats = michael_jordan_advanced_stats[['Season', 'WS', 'BPM', 'VORP', 'PER', 'USG%']]
michael_jordan_selected_total_stats = michael_jordan_total_stats[['Season', 'Trp-Dbl']]
michael_jordan_selected_regular_stats
michael_jordan_selected_advanced_stats
michael_jordan_selected_total_stats
```

```python
michael_jordan_comb = michael_jordan_selected_regular_stats.merge(michael_jordan_selected_total_stats, on='Season', how='left')
michael_jordan_comb
michael_jordan_combined = michael_jordan_comb.merge(michael_jordan_selected_advanced_stats, on='Season', how='left')
michael_jordan_combined
```

```python
jordan_round1 = ["1984-85", "1986-87", "1987-88", "1988-89", "1989-90", "1990-91", "1991-92", "1992-93", "1995-96", "1996-97", "1997-98", "2001-02", "2002-03"]
jordan_round1_seasons = pd.merge(michael_jordan_comb, michael_jordan_selected_advanced_stats, on='Season')
jordan_round1_seasons_combined = jordan_round1_seasons[jordan_round1_seasons["Season"].isin(jordan_round1)]
jordan_round1_seasons_combined
jordan_round2 = ["1984-85", "1986-87", "1987-88", "1988-89", "1989-90", "1990-91", "1991-92", "1992-93", "1995-96", "1996-97", "1997-98"]
jordan_round2_seasons = pd.merge(michael_jordan_comb, michael_jordan_selected_advanced_stats, on='Season')
jordan_round2_seasons_combined = jordan_round2_seasons[jordan_round2_seasons["Season"].isin(jordan_round2)]
jordan_round2_seasons_combined
```

```python
plt.figure(figsize=(12, 6))
ax = sns.barplot(data=jordan_round2_seasons_combined, x='Season', y='PTS', hue='PTS')
plt.xticks(rotation=45, ha="right")
plt.title('Michael Jordan PPG')
plt.xlabel('Season')
plt.ylabel('Metric Count')
plt.legend(title="Metric", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
```

```python
melted_jordan_advanced_data1 = jordan_round2_seasons_combined.melt(
    id_vars=['Season'], 
    value_vars=['WS', 'BPM', 'VORP'],
    var_name='Metric', 
    value_name='Count'
)
```

```python
plt.figure(figsize=(12, 6))
ax = sns.barplot(
data=melted_jordan_advanced_data1, x='Season', y='Count', hue= 'Metric', palette="coolwarm")
plt.xticks(rotation=45, ha="right")
plt.title('Michael Jordan Win Shares')
plt.xlabel('Season')
plt.ylabel('Metric Count')
plt.legend(title="Metric", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
jordan_round3 = ["1986-87", "1987-88", "1988-89", "1989-90", "1990-91", "1991-92", "1992-93", "1995-96", "1996-97"]
jordan_round3_seasons = pd.merge(michael_jordan_comb, michael_jordan_selected_advanced_stats, on='Season')
jordan_round3_seasons_combined = jordan_round3_seasons[jordan_round3_seasons["Season"].isin(jordan_round3)]
jordan_round3_seasons_combined
```

```python
plt.figure(figsize=(10, 6))
scatter = sns.scatterplot(x=jordan_round3_seasons_combined["PTS"], y=jordan_round3_seasons_combined["WS"], color="red", s=100, edgecolor="black", alpha=0.7)


for i, row in jordan_round3_seasons_combined.iterrows():
    plt.text(row["PTS"], row["WS"], row["Season"], fontsize=10, ha="right", color="black")


plt.xlim(25, jordan_round3_seasons_combined["PTS"].max() + 1) 
plt.ylim(15, jordan_round3_seasons_combined["WS"].max() + 1)  


plt.title("Scatter Plot of PPG vs. WS", fontsize=14, color="black")
plt.xlabel("PPG (Points Per Game)", fontsize=12, color="black")
plt.ylabel("WS (Win Shares)", fontsize=12, color="black")
plt.grid(True, linestyle="--", alpha=0.6)
```

```python
jordan_round4 = ["1986-87", "1987-88", "1988-89", "1989-90", "1990-91", "1995-96"]
jordan_round4_seasons = pd.merge(michael_jordan_comb, michael_jordan_selected_advanced_stats, on='Season')
jordan_round4_seasons_combined = jordan_round4_seasons[jordan_round4_seasons["Season"].isin(jordan_round4)]
jordan_round4_seasons_combined
```

```python
df = jordan_round4_seasons_combined.set_index("Season")
fig, ax = plt.subplots(figsize=(12, 6))
df[["TRB", "AST", "STL", "BLK"]].plot(kind="bar", stacked=True, colormap="Reds", edgecolor="black", ax=ax)
ax.set_xticks(range(len(df.index)))
ax.set_xticklabels(df.index, rotation=45, ha="right", fontsize=10)
ax.set_title("Stacked Column Chart of TRB, AST, STL, and BLK", fontsize=14)
ax.set_xlabel("Season", fontsize=12)
ax.set_ylabel("Metric Count", fontsize=12)
ax.legend(title="Metric", loc="upper left")
ax.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```
